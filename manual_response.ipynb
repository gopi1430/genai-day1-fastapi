```python
from fastapi import FastAPI
from pydantic import BaseModel
import uvicorn
import nest_asyncio

```

```python
api = FastAPI(
    title = "manual api",
    description = "a simple api using fastapi" ,
    version = "1.0"
)
print("your fast api has been created ")
```

```python
class GenerateRequest(BaseModel):
    text: str
def gen_response(text):
        return f"AI got your request:{text}"
result = gen_response("what is api")
print(result)
```

```python
@api.post("/generate")
def generate(request : GenerateRequest):
    response = gen_response(request.text)
    return {
        "input" : request.text ,
        "response": response
    }
nest_asyncio.apply()
```

```python
import threading
config = uvicorn.Config(
    api,
    host = "127.0.0.1" ,
    port = "8001" ,
    log_level= "info"
)
server = uvicorn.Server(config)
thread = threading.Thread(target = server.run)
thread.start()
```
